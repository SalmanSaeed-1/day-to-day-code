```python
class bankaccount:
    def __init__(self , account_holder , account_number , balance):
        self.account_holder = account_holder
        self.account_number = account_number
        self.__balance = balance
        
    def deposit(self , amount):
        self.__balance += amount
    
    def withdraw(self , amount):
        self.__balance -= amount
    
    def show_bal(self):
        print("Balance : " , self.__balance)
    
    def disp_acc(self):
        print("Account holder : " , self.account_holder)
        print("Account number : " , self.account_number)
        
        
```

```python
b1 = bankaccount("Salman" , 112 , 5000)
b1.deposit(5000)
b1.withdraw(2000)
b1.show_bal()
b1.disp_acc()
```

```python
class employee:
    def __init__(self, name, emp_id, salary):
        self.name = name
        self.emp_id = emp_id
        self.salary = salary

    def showdetails(self):
        print("Name:", self.name)
        print("Employee id:", self.emp_id)
        print("Salary:", self.salary)

    def bonus(self):
        amount = (self.salary / 100) * 10
        return amount


class manager(employee):
    def __init__(self, name, emp_id, salary, department):
        super().__init__(name, emp_id, salary)
        self.department = department

    def showdetails(self):
        super().showdetails()
        print("Department:", self.department)

    def bonus(self):
        amount = (self.salary / 100) * 20
        return amount


emp1 = employee("Ali", 101, 50000)
m1 = manager("Sara", 102, 80000, "IT")

emp1.showdetails()
print("Bonus:", emp1.bonus())

print()

m1.showdetails()
print("Bonus:", m1.bonus())
```

```python
emails = ["user1@gmail.com", "admin@company.org"]   # List of email addresses

email_dict = {}   # Dictionary to store username and domain pairs

for email in emails:    # Iterate through each email address
    parts = email.split("@")   # Split the email address into username and domain using "@" 
    username = parts[0]    #    Extract the username (the part before "@")
    domain = parts[1]   # Extract the domain (the part after "@")
    email_dict[username] = domain   # Add the username and domain to the dictionary, with the username as the key and the domain as the value
    
print(email_dict)
print(parts)
```
